```python

import requests
import pandas as pd
from bs4 import BeautifulSoup
from pathlib import Path
from datetime import datetime

# Create output folder
Path("output").mkdir(exist_ok=True)

# Read our pilot dataset
pilot_df = pd.read_csv("data/pilot_research.csv")

# Function to fetch and extract webpage content
def fetch_documentation(app_name, url):

    result = {
        "Application": app_name,
        "Source_URL": url,
        "Fetch_Status": "",
        "Page_Title": "",
        "Extracted_Text": "",
        "Checked_At": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    try:
        headers = {
            "User-Agent": "Mozilla/5.0 (compatible; ResearchBot/1.0)"
        }

        response = requests.get(
            url,
            headers=headers,
            timeout=25
        )

        result["Fetch_Status"] = response.status_code

        soup = BeautifulSoup(response.text, "html.parser")

        # Extract page title
        if soup.title:
            result["Page_Title"] = soup.title.get_text(strip=True)

        # Remove unnecessary elements
        for element in soup(["script", "style", "nav", "footer"]):
            element.decompose()

        # Extract readable text
        page_text = soup.get_text(separator=" ", strip=True)

        result["Extracted_Text"] = page_text[:5000]

    except Exception as error:
        result["Fetch_Status"] = "Error"
        result["Extracted_Text"] = str(error)

    return result


# Fetch documentation for both applications
results = []

for _, row in pilot_df.iterrows():

    app_name = row["Application"]
    url = row["API_Documentation_URL"]

    print("Fetching:", app_name)

    result = fetch_documentation(app_name, url)
    results.append(result)


# Convert results into a DataFrame
research_results = pd.DataFrame(results)

# Save extracted research evidence
research_results.to_csv(
    "output/pilot_source_checks.csv",
    index=False
)

print("\nResearch fetching completed!")
display(research_results[
    ["Application", "Fetch_Status", "Page_Title"]
])
```

```python

import pandas as pd
import re

# Load fetched documentation
source_df = pd.read_csv(
    "output/pilot_source_checks.csv",
    dtype=str
).fillna("")

# Keyword groups
keyword_groups = {
    "OAuth": ["oauth", "oauth 2.0"],
    "API Key": ["api key", "apikey"],
    "Access Token": ["access token", "bearer token"],
    "REST API": ["rest api", "restful api"],
    "GraphQL": ["graphql"],
    "Webhooks": ["webhook", "webhooks"],
    "MCP": ["model context protocol", "mcp server"]
}

# Classify documentation text
def identify_keywords(text):

    text = text.lower()
    findings = []

    for category, keywords in keyword_groups.items():

        if any(keyword in text for keyword in keywords):
            findings.append(category)

    return ", ".join(findings) if findings else "No keyword detected"


# Apply classification
source_df["Detected_Research_Signals"] = (
    source_df["Extracted_Text"].apply(identify_keywords)
)

# Save classification results
source_df.to_csv(
    "output/pilot_classification.csv",
    index=False
)

print("Keyword classification completed!")

display(
    source_df[
        ["Application", "Fetch_Status",
         "Detected_Research_Signals"]
    ]
)
```

```python

import pandas as pd

# Load fetched documentation
source_df = pd.read_csv(
    "output/pilot_source_checks.csv",
    dtype=str
).fillna("")

# Feature-specific keywords
feature_keywords = {
    "OAuth": ["oauth", "oauth 2.0"],
    "API Key": ["api key", "apikey"],
    "Access Token": ["access token", "bearer token"],
    "REST API": ["rest api", "restful api"],
    "GraphQL": ["graphql"],
    "Webhooks": ["webhook", "webhooks"],
    "MCP": ["model context protocol", "mcp server"]
}

results = []

for _, row in source_df.iterrows():

    app = row["Application"]
    text = row["Extracted_Text"].lower()
    status = str(row["Fetch_Status"]).strip()

    for feature, keywords in feature_keywords.items():

        if status != "200" or not text.strip():
            prediction = "Unable to Assess"

        elif any(keyword in text for keyword in keywords):
            prediction = "Signal Detected"

        else:
            prediction = "Not Detected"

        results.append({
            "Application": app,
            "Feature": feature,
            "Automated_Prediction": prediction,
            "Fetch_Status": status,
            "Evidence_Source": row["Source_URL"],
            "Interpretation": (
                "Keyword found in fetched text"
                if prediction == "Signal Detected"
                else "No keyword found; absence not established"
                if prediction == "Not Detected"
                else "Source could not be fetched"
            )
        })

# Create improved results table
improved_df = pd.DataFrame(results)

# Save results
improved_df.to_csv(
    "output/improved_classification.csv",
    index=False
)

print("Improved classification completed!")
print("\nPrediction summary:")
print(improved_df["Automated_Prediction"].value_counts())

display(improved_df)
```

```python
import pandas as pd

# Load the master dataset
master_df = pd.read_csv("data/apps_100.csv", dtype=str).fillna("")

# Fields required for research
required_fields = [
    "Authentication_Methods",
    "Access_Model",
    "API_Type",
    "API_Capabilities",
    "Existing_MCP",
    "Buildability_Verdict",
    "API_Documentation_URL"
]

# Calculate completed and pending fields for every app
for field in required_fields:
    master_df[field] = master_df[field].astype(str).str.strip()

master_df["Completed_Fields"] = (
    master_df[required_fields] != ""
).sum(axis=1)

master_df["Research_Progress_%"] = (
    master_df["Completed_Fields"] / len(required_fields) * 100
).round(1)

# Category-wise summary
category_summary = master_df.groupby("Category").agg(
    Total_Apps=("Application", "count"),
    Apps_With_Research=("Completed_Fields", lambda x: (x > 0).sum()),
    Average_Progress=("Research_Progress_%", "mean")
).reset_index()

category_summary["Average_Progress"] = (
    category_summary["Average_Progress"].round(1)
)

# Overall summary
print("OVERALL RESEARCH COVERAGE")
print("-------------------------")
print("Total Applications:", len(master_df))
print("Applications with some research:",
      (master_df["Completed_Fields"] > 0).sum())
print("Fully researched applications:",
      (master_df["Completed_Fields"] == len(required_fields)).sum())
print("Average research progress:",
      round(master_df["Research_Progress_%"].mean(), 1), "%")

print("\nCATEGORY-WISE RESEARCH COVERAGE")
display(category_summary)

# Save reports
category_summary.to_csv(
    "output/category_research_coverage.csv", index=False
)

master_df.to_csv(
    "output/research_progress_100_apps.csv", index=False
)

print("\nBoth coverage reports saved successfully!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

master_df = pd.read_csv(
    "data/apps_100.csv",
    dtype=str
).fillna("")

# Count apps with at least one researched field
research_fields = [
    "Authentication_Methods",
    "Access_Model",
    "API_Type",
    "API_Capabilities",
    "Existing_MCP",
    "Buildability_Verdict"
]

master_df["Research_Fields_Completed"] = (
    master_df[research_fields]
    .apply(lambda row: sum(value.strip() != "" for value in row), axis=1)
)

master_df["Progress_Percentage"] = (
    master_df["Research_Fields_Completed"] / len(research_fields) * 100
)

category_progress = master_df.groupby("Category").agg(
    Total_Apps=("Application", "count"),
    Apps_With_Research=(
        "Research_Fields_Completed",
        lambda x: (x > 0).sum()
    ),
    Average_Progress=("Progress_Percentage", "mean")
).reset_index()

category_progress["Average_Progress"] = (
    category_progress["Average_Progress"].round(1)
)

display(category_progress)

# Create chart
plt.figure(figsize=(12, 6))
plt.barh(
    category_progress["Category"],
    category_progress["Average_Progress"]
)
plt.xlabel("Research Progress (%)")
plt.ylabel("Application Category")
plt.title("Category-wise Research Progress")
plt.xlim(0, 100)
plt.tight_layout()

plt.savefig(
    "output/category_progress.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Dashboard saved successfully!")
```

```python

import pandas as pd
import requests
from bs4 import BeautifulSoup
from datetime import datetime
from pathlib import Path

Path("output").mkdir(exist_ok=True)

source_df = pd.read_csv(
    "data/next_batch_sources.csv",
    dtype=str
).fillna("")

def fetch_page(app, url):
    result = {
        "Application": app,
        "Source_URL": url,
        "Fetch_Status": "",
        "Page_Title": "",
        "Extracted_Text": "",
        "Checked_At": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    try:
        headers = {
            "User-Agent": "Mozilla/5.0"
        }

        response = requests.get(
            url,
            headers=headers,
            timeout=25
        )

        result["Fetch_Status"] = response.status_code

        soup = BeautifulSoup(response.text, "html.parser")

        if soup.title:
            result["Page_Title"] = soup.title.get_text(strip=True)

        for element in soup(["script", "style", "nav", "footer"]):
            element.decompose()

        result["Extracted_Text"] = soup.get_text(
            separator=" ",
            strip=True
        )[:5000]

    except Exception as error:
        result["Fetch_Status"] = "Error"
        result["Extracted_Text"] = str(error)

    return result


results = []

for _, row in source_df.iterrows():

    app = row["Application"]

    urls = [
        row["API_Documentation_URL"],
        row["Authentication_Evidence_URL"]
    ]

    for url in urls:
        print("Fetching:", app, url)
        results.append(fetch_page(app, url))


fetch_df = pd.DataFrame(results)

fetch_df.to_csv(
    "output/next_batch_source_checks.csv",
    index=False
)

display(
    fetch_df[
        ["Application", "Fetch_Status", "Page_Title"]
    ]
)

print("Documentation fetching completed!")
```

```python

import pandas as pd

fetch_df = pd.read_csv(
    "output/next_batch_source_checks.csv",
    dtype=str
).fillna("")

feature_keywords = {
    "OAuth 2.0": ["oauth 2.0", "oauth"],
    "API Key / Token": ["api token", "api key", "apikey"],
    "Access Token": ["access token", "bearer token"],
    "Refresh Token": ["refresh token"],
    "REST API": ["rest api", "restful", "api endpoint"],
    "GraphQL": ["graphql"],
    "Webhooks": ["webhook", "webhooks"],
    "MCP": ["model context protocol", "mcp server"]
}

results = []

for _, row in fetch_df.iterrows():

    text = row["Extracted_Text"].lower()
    status = str(row["Fetch_Status"])

    for feature, keywords in feature_keywords.items():

        if status != "200" or not text.strip():
            prediction = "Unable to Assess"
            matched = ""
        else:
            matched = [
                word for word in keywords if word in text
            ]

            prediction = (
                "Signal Detected" if matched
                else "Not Detected"
            )

        results.append({
            "Application": row["Application"],
            "Feature": feature,
            "Fetch_Status": status,
            "Automated_Prediction": prediction,
            "Matched_Keywords": ", ".join(matched),
            "Interpretation": (
                "Keyword found in fetched page"
                if prediction == "Signal Detected"
                else "No keyword found; absence not established"
                if prediction == "Not Detected"
                else "Source could not be assessed"
            )
        })

classification_df = pd.DataFrame(results)

classification_df.to_csv(
    "output/next_batch_classification.csv",
    index=False
)

display(classification_df)

print("\nPrediction Summary:")
print(classification_df["Automated_Prediction"].value_counts())

print("\nClassification file saved!")
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from concurrent.futures import ThreadPoolExecutor, as_completed

df = pd.read_csv("data/apps_100.csv").fillna("")
source_cols = [
    "API_Documentation_URL",
    "Authentication_Evidence_URL",
    "MCP_Evidence_URL"
]

jobs = []

for _, row in df.iterrows():
    for col in source_cols:
        url = str(row.get(col, "")).strip()
        if url.startswith("http"):
            jobs.append((row["Application"], row["Category"], col, url))

def fetch_source(job):
    app, category, source_type, url = job
    try:
        r = requests.get(
            url,
            headers={"User-Agent": "Mozilla/5.0"},
            timeout=12
        )
        soup = BeautifulSoup(r.text, "html.parser")
        return {
            "Application": app,
            "Category": category,
            "Source_Type": source_type,
            "URL": url,
            "Fetch_Status": r.status_code,
            "Page_Title": soup.title.get_text(strip=True) if soup.title else "",
            "Extracted_Text": soup.get_text(" ", strip=True)[:2500]
        }
    except Exception as e:
        return {
            "Application": app,
            "Category": category,
            "Source_Type": source_type,
            "URL": url,
            "Fetch_Status": "Error",
            "Page_Title": str(e),
            "Extracted_Text": ""
        }

results = []
with ThreadPoolExecutor(max_workers=8) as executor:
    futures = [executor.submit(fetch_source, j) for j in jobs]
    for future in as_completed(futures):
        results.append(future.result())

audit = pd.DataFrame(results)
audit.to_csv("output/all_100_source_audit.csv", index=False)

print("Total source URLs:", len(jobs))
print("Fetched:", len(audit))
print("\nHTTP status summary:")
print(audit["Fetch_Status"].value_counts().to_string())

print("\nCategory-wise source coverage:")
print(audit.groupby("Category")["Application"].nunique().to_string())

print("\nSaved: output/all_100_source_audit.csv")
```

```python
import pandas as pd

df = pd.read_csv("data/apps_100.csv").fillna("")

source_cols = [
    "API_Documentation_URL",
    "Authentication_Evidence_URL",
    "MCP_Evidence_URL"
]

# App-level source coverage
for col in source_cols:
    df[col + "_Present"] = df[col].astype(str).str.startswith("http")

df["Sources_Available"] = df[[c + "_Present" for c in source_cols]].sum(axis=1)

# Category summary
summary = df.groupby("Category").agg(
    Total_Apps=("Application", "count"),
    Apps_With_Sources=("Sources_Available", lambda x: (x > 0).sum()),
    Total_Source_URLs=("Sources_Available", "sum")
).reset_index()

summary["Apps_Pending"] = summary["Total_Apps"] - summary["Apps_With_Sources"]

# Pending apps
pending = df[df["Sources_Available"] == 0][
    ["Application", "Category", "Official_Website"]
]

summary.to_csv("output/category_research_gaps.csv", index=False)
pending.to_csv("output/pending_apps_research.csv", index=False)

print("CATEGORY-WISE RESEARCH GAPS")
display(summary)

print("\nAPPS WITHOUT SOURCE URLS:", len(pending))
display(pending)

print("\nReports saved successfully!")
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from urllib.parse import quote
import time

df = pd.read_csv("data/apps_100.csv").fillna("")
results = []

for _, row in df.iterrows():
    app = row["Application"]
    category = row["Category"]

    query = quote(f"{app} official API documentation developer")
    url = f"https://html.duckduckgo.com/html/?q={query}"

    try:
        r = requests.get(
            url,
            headers={"User-Agent": "Mozilla/5.0"},
            timeout=15
        )
        soup = BeautifulSoup(r.text, "html.parser")

        links = soup.select(".result__a")

        for link in links[:5]:
            results.append({
                "Application": app,
                "Category": category,
                "Candidate_Title": link.get_text(" ", strip=True),
                "Candidate_URL": link.get("href", ""),
                "Verification_Status": "Needs manual verification"
            })

    except Exception as e:
        results.append({
            "Application": app,
            "Category": category,
            "Candidate_Title": "Search failed",
            "Candidate_URL": str(e),
            "Verification_Status": "Not verified"
        })

    time.sleep(1)

research = pd.DataFrame(results)
research.to_csv("output/100_app_api_candidates.csv", index=False)

print("Apps processed:", df["Application"].nunique())
print("Candidate links found:", len(research))
print("Saved: output/100_app_api_candidates.csv")
display(research.head(25))
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

path = "data/apps_100.csv"
df = pd.read_csv(path).fillna("")
apps = df[df["Category"] == "AI, Research & Media"].copy()

results = []

for _, row in apps.iterrows():
    for kind, col in [
        ("API", "API_Documentation_URL"),
        ("Auth", "Authentication_Evidence_URL")
    ]:
        url = row.get(col, "")
        try:
            r = requests.get(
                url, timeout=12,
                headers={"User-Agent": "Mozilla/5.0"}
            )
            soup = BeautifulSoup(r.text, "html.parser")
            results.append({
                "Application": row["Application"],
                "Source_Type": kind,
                "URL": url,
                "HTTP_Status": r.status_code,
                "Page_Title": soup.title.get_text(strip=True) if soup.title else "",
                "Fetch_Result": "Accessible" if r.status_code < 400 else "Blocked/Failed"
            })
        except Exception as e:
            results.append({
                "Application": row["Application"],
                "Source_Type": kind,
                "URL": url,
                "HTTP_Status": "Error",
                "Page_Title": "",
                "Fetch_Result": str(e)[:120]
            })

audit = pd.DataFrame(results)
audit.to_csv("output/ai_source_audit.csv", index=False)

# Create a concise gap report
gaps = []
for _, row in apps.iterrows():
    gaps.append({
        "Application": row["Application"],
        "API_Source": "Added" if row.get("API_Documentation_URL") else "Missing",
        "Auth_Source": "Added" if row.get("Authentication_Evidence_URL") else "Missing",
        "Research_Status": row.get("Research_Status", "Pending"),
        "MCP_Status": row.get("Existing_MCP", "Needs verification")
    })

gap_report = pd.DataFrame(gaps)
gap_report.to_csv("output/ai_evidence_gap_report.csv", index=False)

display(audit[["Application", "Source_Type", "HTTP_Status", "Fetch_Result"]])
display(gap_report)

print("AI source audit and gap report saved!")
```
